# **3. Resolución de los problemas de calidad**

---

```python
print("\n------------------ (1. Nulos en transaction_type) ------------------")
print("Columna transaction_type nulos antes")
print(ventas["transaction_type"].value_counts(dropna=False))#Cantidad de cada valor, incluyendo los nulos

ventas["transaction_type"] = ventas["transaction_type"].fillna("No registrado")#fillna(): Reemplaza los valores nulos por "No registrado"

print("\nColumna transaction_type nulos despues")
print(ventas["transaction_type"].value_counts(dropna=False))
print("---------------------------------------------------")
```

```python
print("\n------ (2. Inconsistencia en formato fecha) -------")# Cuenta las fechas que cumplen con el formato MM/DD/YYYY antes de corregirlas
print("Cantidad de fechas con formato correcto antes (MM/DD/YY): ", ventas["date"].str.match(r"^\d{2}/\d{2}/\d{4}$").sum())
print("Tipo de dato de fecha antes: ",ventas["date"].dtypes)#dtypes: tipo de dato de la fecha

ventas["date"] = pd.to_datetime(ventas["date"], format="mixed")# pd.to_datetime(): Convierte valores a formato de fecha y hora
#format="mixed": Permite convertir fechas con diferentes formatos.

print("\nCantidad de fechas con formato correcto despues (MM/DD/YY): ", ventas["date"].dt.strftime("%m/%d/%Y").str.match(r"^\d{2}/\d{2}/\d{4}$").sum())
print("Tipo de dato de fecha despues: ",ventas["date"].dtypes)
print("---------------------------------------------------")#strftime(): Convierte fechas a un formato de texto
#str.match(): Comprueba si los valores coinciden con un patrón.
```

```python
print("\n------ (3. Espacio en memoria) -------")
print("Espacio de memoria antes")
display(pd.DataFrame({
    "Memoria (bytes)": ventas.memory_usage(deep=True, index=False),#Muestra la memoria utilizada por cada columna, en bytes
    "Valores únicos": ventas.nunique(),#nunique() → Cuenta la cantidad de valores únicos por columna
    "Tipo": ventas.dtypes#dtypes:tipo de datos
}, dtype=object))#dtype=object: convertierte el df a tipo objeto, para que los valores unicos sean enteros y no decimales


columnas_categoria = ["item_name", "item_type", "transaction_type", "received_by", "time_of_sale"]#lista de columnas a cambiar a categoria
ventas[columnas_categoria] = ventas[columnas_categoria].astype("category")# Convierte las columnas seleccionadas al tipo categoria

print("\nEspacio de memoria Despues")
display(pd.DataFrame({
    "Memoria (bytes)": ventas.memory_usage(deep=True, index=False),
    "Valores únicos": ventas.nunique(),
    "Tipo": ventas.dtypes
}, dtype=object))

print("---------------------------------------------------")
```

**Problemas de calidad resueltos**
* **1. Nulos**
    * Se reemplazaron los valores nulos de transaction_type por "No registrado", ya que representan el 10.7% de los registros. No se utilizó la moda porque Cash (476) y Online (417) tienen cantidades similares, y asignar los 107 nulos a Cash aumentaría artificialmente la diferencia entre ambas categorías.
* **2. Inconsistencias**
    * Se convirtió la columna date a formato datetime, ya que presentaba diferentes formas de representar las fechas, como meses con uno o dos dígitos. Después de la conversión, los 1,000 registros cumplen con el formato MM/DD/YYYY.
* **3. Memoria**
    * Se convirtieron las columnas item_name, item_type, transaction_type, received_by y time_of_sale al tipo category, ya que contienen pocos valores únicos que se repiten en muchos registros. Esto permite reducir el espacio de memoria utilizado por el DataFrame sin modificar los valores de los datos.

# **4. Traducción de datos a español**

---

```python
print("\n---------------- (1. Valores originales en ingles) -----------------")
print("Item type:", ventas["item_type"].unique())
print("Item name:", ventas["item_name"].unique())
print("Time of sale:", ventas["time_of_sale"].unique())
print("Transaction type:", ventas["transaction_type"].unique())
print("---------------------------------------------------")
```

```python
#Listas para traducir los valores a español
TraduccionColumnas = {
    "order_id": "ID de orden",
    "date": "Fecha",
    "item_name": "Nombre del producto",
    "item_type": "Tipo de producto",
    "item_price": "Precio del producto",
    "quantity": "Cantidad",
    "transaction_amount": "Monto de la transacción",
    "transaction_type": "Medio de pago",
    "received_by": "Recibido por",
    "time_of_sale": "Turno de venta"
}

TraduccionItemName = {
    "Aalopuri": "Aalopuri",            #Sin traduccion adecuada al espanol, se mantiene el nombre original (plato tipico indio)
    "Vadapav": "Vadapav",              #Sin traduccion adecuada al espanol, se mantiene el nombre original (plato tipico indio)
    "Sugarcane juice": "Jugo de caña",
    "Panipuri": "Panipuri",            #Sin traduccion adecuada al espanol, se mantiene el nombre original (plato tipico indio)
    "Frankie": "Frankie",              #Nombre propio del producto, se mantiene igual
    "Sandwich": "Emparedado",
    "Cold coffee": "Café frío"
}

TraduccionItemType = {
    "Fastfood": "Comida rápida",
    "Beverages": "Bebidas"
}

TraduccionTransactionType = {
    "Cash": "Efectivo",
    "Online": "En línea",
    "No registrado": "No registrado"  #El valor ya se encuentra en español
}

TraduccionReceivedBy = {
    "Mr.": "Sr.",
    "Mrs.": "Sra."
}
TraduccionTimeOfSale = {
    "Morning": "Mañana",
    "Afternoon": "Tarde",
    "Evening": "Anochecer",
    "Night": "Noche",
    "Midnight": "Medianoche"
}

print("\n---------------- (2. Valores traducidos al español) -----------------")
print("Columnas:", ventas.rename(columns=TraduccionColumnas).columns.tolist())
print("Nombre de Producto: ",ventas["item_name"].map(TraduccionItemName).astype(str).unique())
print("Categoria de Producto: ",ventas["item_type"].map(TraduccionItemType).astype(str).unique())
print("Tipo de transaccion: ",ventas["transaction_type"].map(TraduccionTransactionType).astype(str).unique())
print("Atendido por: ",ventas["received_by"].map(TraduccionReceivedBy).astype(str).unique())
print("Tiempo de Venta: ",ventas["time_of_sale"].map(TraduccionTimeOfSale).astype(str).unique())
print("---------------------------------------------------")
```

**Justificación de la traducción**  
* En item_name, **Aalopuri**, **Vadapav**, **Panipuri** y **Frankie** son platillos de comida callejera india sin una traducción adecuada al español (no existe una palabra equivalente que describa el platillo); se decidió mantener el nombre original, ya que traducirlos literalmente (por ejemplo, palabra por palabra) perdería el significado del producto. Solo **Sugarcane juice** y **Cold coffee** tenían una traducción natural (Jugo de caña, Café frío) y se tradujeron.
* Los nombres de tipo de producto (item_type), turno (time_of_sale) y medio de pago (transaction_type) tienen equivalente directo en español y se tradujeron por completo.

### **1. Reporte por medio de pago**

```python
mediosPago = ventasLimpio[ventasLimpio["transaction_type"].isin(["Cash", "Online"])].groupby("transaction_type").agg(#groupby(): agrupa por medio de pago
    Ventas_Totales=("order_id", "count"),#isin(): selecciona Cash y Online
    Monto_Total=("transaction_amount", "sum")
)

mediosPago["(%)_Ventas"] = mediosPago["Ventas_Totales"] / len(ventasLimpio) * 100# Calcula el porcentaje de ventas
mediosPago["(%)_Monto"] = mediosPago["Monto_Total"] / ventasLimpio["transaction_amount"].sum() * 100# Calcula el porcentaje del monto
mediosPago.index.name = TraduccionColumnas["transaction_type"]# Traduce los resultados

print("-------------------- (1. Reporte por medio de pago) -------------------")
display(mediosPago.rename(index=TraduccionTransactionType).round(2))
print("-----------------------------------------------------------------------")
```

**Hallazgos**
* Efectivo: 476 ventas y 47.6% del total.
* En línea: 417 ventas y 41.7% del total.
* Efectivo generó mayor monto total: 132,840.

**Método de detección**
* isin(), groupby() y agg() para filtrar, agrupar y calcular.
* count para ventas y sum para montos.

### **2. Reporte de productos por turno**

```python
productos = ["Cold coffee", "Panipuri", "Sandwich"]

cantidadProductos = ventasLimpio[ventasLimpio["item_name"].isin(productos)].pivot_table(#pivot_table(): crea una tabla resumen
    index="item_name",
    columns="time_of_sale",
    values="quantity",
    aggfunc="sum",
    observed=True,
    fill_value=0
)

cantidadProductos = cantidadProductos.reindex(columns=["Morning", "Afternoon", "Evening", "Night", "Midnight"])#Organiza los turnos en orden
cantidadProductos = cantidadProductos.rename(index=TraduccionItemName,columns=TraduccionTimeOfSale)
cantidadProductos.index.name = TraduccionColumnas["item_name"]# rename(): traduce nombres en la salida
cantidadProductos.columns.name = TraduccionColumnas["time_of_sale"]

print("------------ (2. Cantidad de productos vendidos por turnos) -----------")
display(cantidadProductos)
print("-----------------------------------------------------------------------")
```

**Hallazgos**
* Café frío: mayor cantidad en la mañana (287).
* Café frío: mayor cantidad en la tarde (337).
* Panipuri: mayor cantidad en anochecer (278).
* Emparedado: mayor cantidad en la noche (292).
* Panipuri: mayor cantidad en medianoche (278).

**Método de detección**
* isin() para seleccionar los productos.
* pivot_table() y sum para obtener las cantidades por turno.
* reindex() para ordenar los turnos.

### **3. Reporte de tipo de producto por medio de pago**

```python
tipoProductoMedioPago = pd.pivot_table(
    ventasLimpio,
    index="item_type",
    columns="transaction_type",
    values="order_id",
    aggfunc="count",
    observed=True,
    fill_value=0
)

tipoProductoMedioPago = tipoProductoMedioPago.rename(index=TraduccionItemType,columns=TraduccionTransactionType)
tipoProductoMedioPago.index.name = TraduccionColumnas["item_type"]# rename(): traduce los nombres en la salida
tipoProductoMedioPago.columns.name = TraduccionColumnas["transaction_type"]

print("-- (3. Cantidad de ventas de cada tipo de producto por medio de pago) --")
display(tipoProductoMedioPago)
print("-----------------------------------------------------------------------")
```

**Hallazgos**
* Comida rápida tiene más ventas: 328 en efectivo y 291 en línea.
* Bebidas tiene 148 ventas en efectivo y 126 en línea.
* Hay 107 ventas sin medio de pago registrado.

Método de detección
**bold text*** pivot_table() para cruzar tipo de producto y medio de pago.
* count para contar las ventas.
* fill_value=0 para reemplazar valores vacíos.

### **4. Reporte de producto más vendido**

```python
unidadesVendidas = pd.pivot_table(
    ventasLimpio,
    index="item_name",
    columns=ventasLimpio["date"].dt.month,# dt.month: obtiene el número del mes
    values="quantity",
    aggfunc="sum",# sum: suma las unidades vendidas
    fill_value=0# Reemplaza los valores vacíos por 0
).reindex(columns=range(1, 13), fill_value=0)# reindex(): asegura los 12 meses

ingresosGenerados = pd.pivot_table(
    ventasLimpio,
    index="item_name",
    columns=ventasLimpio["date"].dt.month,# dt.month: obtiene el número del mes
    values="transaction_amount",
    aggfunc="sum",
    fill_value=0
).reindex(columns=range(1, 13), fill_value=0)# reindex(): asegura los 12 meses

reporte = pd.DataFrame({
    "Producto más vendido": unidadesVendidas.idxmax(),# idxmax(): obtiene el producto con mayor cantidad
    "Unidades vendidas": unidadesVendidas.max(),# max(): obtiene la cantidad máxima
    "Producto con mayor ingreso": ingresosGenerados.idxmax(),# idxmax(): obtiene el producto con mayor ingreso
    "Ingreso generados": ingresosGenerados.max()# max(): obtiene el ingreso máximo
})
#Ordena los meses en texto
reporte.index = ["Enero", "Febrero", "Marzo", "Abril", "Mayo", "Junio","Julio", "Agosto", "Septiembre", "Octubre", "Noviembre", "Diciembre"]
reporte.index.name = "Mes"

print("--------------- (4. Producto con más cantidades vendidas y con mayores ingresos por meses de manera global) ---------------")
display(reporte)
print("---------------------------------------------------------------------------------------------------------------------------")
```

**Hallazgos**
* Frankie fue el más vendido en 3 meses.
* Sandwich obtuvo el mayor ingreso en 5 meses.
* Enero tuvo el mayor ingreso: 8,580.

**Método de detección**
* pivot_table() para resumir unidades e ingresos por mes y producto.
* dt.month para obtener el mes.
* idxmax() para identificar el producto con mayor valor.
* max() para obtener la cantidad o ingreso máximo.

### **1. Ingresos trimestrales del año 2022**
¿Cómo se distribuyen los montos totales de las transacciones durante los trimestres del año 2022?

```python
montosTrimestre = pd.pivot_table(
    ventasLimpio[ventasLimpio["date"].dt.year == 2022],# Filtra las ventas de 2022
    index=ventasLimpio["date"].dt.quarter, # Obtiene el trimestre
    values="transaction_amount",
    aggfunc="sum",# Suma los montos
    fill_value=0
).reindex([1, 2, 3, 4], fill_value=0) # Fuerza a incluir los 4 trimestres
# Calcula el porcentaje de cada trimestre
montosTrimestre["Porcentaje"] = (montosTrimestre["transaction_amount"] /montosTrimestre["transaction_amount"].sum() * 100)
#Ordena los trimestres por rangos de meses
montosTrimestre.index = ["Enero - Marzo", "Abril - Junio", "Julio - Septiembre", "Octubre - Diciembre"]

montosTrimestre.index.name = "Trimestre"
montosTrimestre = montosTrimestre.rename(columns={# Traduce el nombre de la columna
    "transaction_amount": "Monto total"
})

print("-------------------- (1. Transacciones por trimestres del año 2022) -------------------")
display(montosTrimestre.round(2))
print("-----------------------------------------------------------------------")
```

**Hallazgos**
* Enero - Marzo: no registró ventas.
* Octubre - Diciembre: tuvo el mayor monto, con $73,565 (37.61%).
* Abril - Junio y Julio - Septiembre tuvieron montos similares.

**Método de comprobación**
* pivot_table() para agrupar los montos por trimestre.
* sum para obtener el monto total.
* Cálculo del porcentaje respecto al total de 2022.

**Decisiones tomadas**
* Se incluyeron los cuatro trimestres aunque uno no tuviera ventas.
* Los trimestres sin ventas se representaron con 0.
* Se utilizaron rangos de meses para facilitar la interpretación.

**Recomendaciones de negocio**
* Investigar y solicitar las causas por que no hay ventas en el primer trimestre.
* Analizar los productos vendidos en el cuarto semestre con mayores ventas.
* Planificar estrategias para aumentar las ventas en el segundo y tercer trimestre del año.

### **2. Participación de ventas de bebidas en el 2023**
qué porcentaje de las ventas de 2023 corresponde a productos de la categoría Bebidas.

# **8. Hallazgos dirigidos al dueño del restaurante**

**Hallazgos**
1. Algunas ventas realizadas no tienen registrado el medio de pago, lo que dificulta conocer con precisión cómo pagan los clientes.
2. La información sobre quién recibe las ventas es poco detallada. Se recomienda registrar el nombre o identificador  del empleado en cada venta.
3. La comida rápida es el punto más fuerte del negocio.
4. El emparedado es el producto estrella del restaurante, ya que, es el  mas ingresos genera.
