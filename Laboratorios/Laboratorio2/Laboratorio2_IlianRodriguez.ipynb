**4. precio_unitario (Media/Mediana) y precio_unitario (Media/Mediana)**

---

```python
print("---------- (4. precio_unitario (Media/Mediana) y precio_unitario (Media/Mediana)) ----------")

columnasNumericas = ["cantidad", "precio_unitario"]

for col in columnasNumericas:

    df[col] = pd.to_numeric(df[col], errors="coerce")#to_numeric: convierte la columna a tipo numérico.
                                                     #coerce: Si encuentra un valor que no puede convertir, lo transforma en NaN.
    mediaCol = df[col].mean()# Calcula la media
    medianaCol = df[col].median()#Calcula la mediana
    sesgo = df[col].skew()#Calcula el sesgo de la distribución

    if abs(sesgo) < 0.5:# Si |skew| < 0.5, se considera una distribución aproximadamente simétrica y se utiliza la media.
        valorImputacion = mediaCol
        criterio = "media"
    else:#Si |skew| >= 0.5, se considera una distribución sesgada y se utiliza la mediana.
        valorImputacion = medianaCol
        criterio = "mediana"

    if col == "cantidad":
        valorImputacion = round(valorImputacion)#cantidad: se redondea a un número entero
    else:
        valorImputacion = round(valorImputacion, 2)#precio_unitario: se redondea a 2 decimales

    df[col] = df[col].fillna(valorImputacion)#fillna:()Reemplaza los valores nulos por media o mediana

    dfMediaMediana = pd.DataFrame({
        "Columna": [col],
        "Sesgo": [round(sesgo, 2)],
        "Criterio": [criterio],
        "Valor imputado": [round(valorImputacion, 2)]
    })

    display(dfMediaMediana)
```

**Hallazgos**  
1. La columna precio_unitario contiene 60 de nulos que representa 1.90% de la columna.
2. La columna cantidad contiene 59 de nulos que representa 1.87% de la columna.
3. Las columnas cantidad y precio_unitario son variables numéricas.

**Método de detección**  
1. Se calculó el sesgo (skewness) de cada columna para determinar si la distribución es aproximadamente simétrica o sesgada.
El criterio utilizado fue:
|sesgo| < 0.5 → se utiliza la media.
|sesgo| >= 0.5 → se utiliza la mediana.  
2. Se utilizó pd.to_numeric() para asegurar que las columnas fueran de tipo numérico.
3. Se utilizó mean() para calcular la media y median() para calcular la mediana.
4. Se utilizó fillna() para reemplazar los valores nulos con el valor seleccionado.

**Decisiones tomadas**  
1. En la columna cantidad se seleccionó automáticamente la media cuando el sesgo fue menor a 0.5 y los Los valores nulos fueron reemplazados utilizando el valor de imputación.
2. En la columna precio_unitario se seleccionó automáticamente la media cuando el sesgo fue menor a 0.5 y los Los valores nulos fueron reemplazados utilizando el valor de imputación.

**2. Reemplazar las inconsistencias por valores estandarizados**

---

```python
print("---------- (2. Reemplazar las inconsistencias por valores estandarizados) ----------")
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].str.replace("Ã¡", "á", regex=False)#str.replace(): reemplaza caracteres incorrectos por su representación correcta
    df[col] = df[col].str.replace("Ã©", "é", regex=False)
    df[col] = df[col].str.replace("Ã­", "í", regex=False)
    df[col] = df[col].str.replace("Ã³", "ó", regex=False)
    df[col] = df[col].str.replace("Ãº", "ú", regex=False)
    df[col] = df[col].str.replace("Ã±", "ñ", regex=False)
    df[col] = df[col].str.replace("Ã\xa0", "à", regex=False)
    df[col] = df[col].str.replace("_", " ", regex=False)

df["promo_producto"] = df["promo_producto"].replace("???", "No especificado")
df["total_venta"] = df["total_venta"].abs()#abs(): elimina el signo negativo convirtiendo el valor a positivo
```

**Hallazgos**
1. Se encontraron errores de codificación en caracteres especiales (Ã¡, Ã©, Ã­, Ã³, Ãº, Ã± y Ã\xa0),
2. Se encontraron guiones bajos (_) en valores de texto para separar palabras.
3. Se encontraron 15 registros con (???) en promo_producto, que no especificaba su funcion.
4. Se encontraron 20 registros con valores negativos en total_venta, que no se especifican que pertenezcan a una devolucion.


**Metodo de deteccion**
1. Use value_counts(dropna=False)) para encontrar los valores unicos de cada columna incluyendo nulos.
2. abs() para convertir los valores negativos a positivos.

**Decisiones tomadas**
1. Se utilizó str.replace() para sustituir las secuencias de caracteres incorrectas por sus caracteres correspondientes (á, é, í, ó, ú, ñ y à).
2. Se utilizó str.replace() para sustituir los guiones bajos (_)por espacios.
3.  Se utilizó str.replace() para sustituir (???)  por No especificado, debido a que no se pudo determinar su significado exacto.
