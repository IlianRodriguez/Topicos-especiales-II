1.	Descargue el siguiente Dataset: https://www.kaggle.com/datasets/datatalesbyagos/retail-hardware-salesventas-ferreteria

```python
import pandas as pd #importo libreria pandas
url = "https://github.com/IlianRodriguez/Topicos-especiales-II/raw/refs/heads/main/Laboratorios/Laboratorio2/retail_ferreteria.csv" #url del csv
df = pd.read_csv(url, encoding="cp1252") #leo el csv y lo guardo en un dataframe inicial
```

2.	Realice el análisis exploratorio (columns, dtypes, head, describe)

```python
print("---------- (1. Columnas del DataFrame) ----------")
display(df.columns.to_frame(index=False, name="Nombre de la columna"))#columns: obtiene los nombres de las columnas. to_frame: convierte el resultado en un DataFrame

print("\n---------- (2. Tipos de datos de las columnas del DataFrame) ----------")
display(df.dtypes.to_frame(name="Tipo de dato"))#dtypes: obtine el tipo de datos de cada columna. to_frame: convierte el resultado en un DataFrame

print("\n---------- (3. Primeras 5 filas del DataFrame) ----------")
display(df.head())#head: muestra las primeras 5 filas de cada columna

print("\n---------- (4. Primeras 5 filas del DataFrame) ----------")
display(df.describe())#describe: Obtener resumen estadistico
```
