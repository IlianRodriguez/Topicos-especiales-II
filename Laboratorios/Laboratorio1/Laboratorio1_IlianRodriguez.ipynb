```python
#1.	Descargue el siguiente Dataset: https://www.kaggle.com/datasets/kyanyoga/sample-sales-data/data
import pandas as pd #importo libreria panadas

url = "https://github.com/IlianRodriguez/Topicos-especiales-II/raw/refs/heads/main/Laboratorios/Laboratorio1/sales_data_sample.csv" #url del csv

df = pd.read_csv(url, encoding="cp1252") #leo el csv y lo guardo
```

```python
#2: Realice el análisis exploratorio
print(df.shape) #muestra filas y columnas
print(df.columns) #muestra los nombres de las columnas
print(df.dtypes) #muestra el tipo de dato de cada columna
print(df.head()) #muestra las primeras 5 filas
print(df.tail()) #muestra las ultimas 5 filas
print(df.describe()) #muestra estadisticas básicas
print(df.info()) #muestra información general del DataFrame
print(df.isnull().sum()) #cuenta los valores nulos por columna
```

```python
#3: Crear un nuevo DataFrame con: ORDERNUMBER, PRODUCTLINE, COUNTRY, CUSTOMERNAME, QUANTITYORDERED, PRICEEACH, SALES, DEALSIZE.
columnas = ["ORDERNUMBER", "PRODUCTLINE", "COUNTRY", "CUSTOMERNAME", "QUANTITYORDERED", "PRICEEACH", "SALES", "DEALSIZE"] #Almaceno las columanas en una variable
df2 = df[columnas] #selecciono las columnas y las guardo en el nuevo dataset
print(df2.columns) #Muestro todas las columnas del nuevo dataframe
```

```python
#4: Filtrar únicamente los pedidos de la línea Vintage Cars.
vintageCars = (df2["PRODUCTLINE"] == "Vintage Cars") #Creo el filtro para vintageCars
dfVintageCars = df2.loc[vintageCars] #Creo el nuevo df y le aplico el filtro vintageCars
print(dfVintageCars) #Muestro solo los pedidos de las filas Vintage Cars
```

```python
#5: Dentro de ese filtro, quedarse solo con los pedidos donde el país sea distinto de USA.
noUSA = (dfVintageCars["COUNTRY"] != "USA") #Dentro del dtVintageCars creo un nuevo filtro para que excluya los pedidos de USA
dfVintageCarsNoUsa = dfVintageCars.loc[noUSA] #Creo un nuevo df teniendo como base dtVintageCars y le agrego el nuevo filtro
print(dfVintageCarsNoUsa) #Muestro solo los pedidos de las filas Vintage Cars que no sean de USA
```

```python
#6: Ordenar ese resultado por SALES de mayor a menor y mostrar los primeros 10.
dfVintageCarsNoUsaSales = dfVintageCarsNoUsa.sort_values(by="SALES", ascending=False)#sort_values para organizar las SALES de mayor a menor
print(dfVintageCarsNoUsaSales.head(10))#head(10) para mostrar las 10 primeras filas
```

```python
#7: Responder con código: ¿qué país aparece con más frecuencia entre esos pedidos?
print(dfVintageCarsNoUsaSales["COUNTRY"].value_counts().head(1)) #value_counts cuanta cuantas veces aparece un pais en la columna COUNTRY y head(1) muestra el primero
```

```python
#8: Comparar df.loc[5] vs df.iloc[5] sobre el DataFrame ya ordenado, y explicar la diferencia en una frase.
print(dfVintageCarsNoUsaSales.loc[5])
#.loc[5] busca el índice 5, que no existe en el DataFrame ordenado, mientras que .iloc[5] busca la sexta fila por su posición.
```

```python
#8: Comparar df.loc[5] vs df.iloc[5] sobre el DataFrame ya ordenado, y explicar la diferencia en una frase.
print(dfVintageCarsNoUsaSales.iloc[5])
#.loc[5] busca el índice 5, que no existe en el DataFrame ordenado, mientras que .iloc[5] busca la sexta fila por su posición.
```

```python
#9: Revisar el tipo de dato de ORDERDATE. ¿Llegó como fecha o como texto?
print(df["ORDERDATE"].dtypes)#resultado de tipo texto
```

```python
#10: Convertirla a fecha con pd.to_datetime().
df["ORDERDATE"] = pd.to_datetime(df["ORDERDATE"]) #to_datetime para forzar la conversión de la columna a fecha
print(df["ORDERDATE"].dtypes) #muestra el tipo de tipo de datos
```

```python
#11: Convertir COUNTRY a tipo category y comparar df.memory_usage(deep=True) antes y después.

#Antes
print("--- Valores de COUNTRY antes ---")
print("Tipo de dato:", df["COUNTRY"].dtype)#tipo object
memoriaAntes = df.memory_usage(deep=True)#memory_usage busca cuánta memoria usa la columnas COUNTRY
memoriaAntesCountry = memoriaAntes["COUNTRY"]#extrae la cantidad de memoria de COUNTRY
print(f"Memoria: {memoriaAntesCountry:,} bytes")#152,568 bytes en uso
print("Valores únicos:", df["COUNTRY"].nunique())#19 valores unicos
print("Total de filas:", df.shape[0])#2823 filas en total

#Despues
print("\n---Valores de COUNTRY Despues---")
df["COUNTRY"] = df["COUNTRY"].astype("category")#Convierte la columna COUNTRY de texto (object) a tipo categórico (category)
print("Tipo de dato:", df["COUNTRY"].dtype)#tipo category
memoriaDespues = df.memory_usage(deep=True)#memory_usage busca cuánta memoria usa la columnas COUNTRY
memoriaDespuesCountry = memoriaDespues["COUNTRY"]#extrae la cantidad de memoria de COUNTRY
print(f"Memoria: {memoriaDespuesCountry:,} bytes")#4,436 bytes en uso
print("Valores únicos:", df["COUNTRY"].nunique())#19 valores unicos
print("Total de filas:", df.shape[0])#2823 filas en total
```

```python
#12.	Realice 4 reportes libres (únicos por cada grupo). Que se encontró, como fue comprobado y recomendación de negocio ( presentar en la siguiente clase).

#1: Año en que se hicieron menos pedidos a Pasadena USA y Cuantos fueron?
PasadenaUSA = (df["COUNTRY"] == "USA") & (df["CITY"] == "Pasadena") # Creo el filtro
dfPasadenaUSA = df.loc[PasadenaUSA] # Selecciono los pedidos de Pasadena, USA
anioMenosPedidos = dfPasadenaUSA["YEAR_ID"].value_counts() # Cuento los pedidos de cada año
print(anioMenosPedidos.sort_values().head(1)) # Muestro el año con menos pedidos y la cantidad
#Se encontró: El año en que se realizaron menos pedidos en Pasadena, USA y la cantidad de pedidos.
#Se comprobó: Se filtraron los pedidos de Pasadena, USA y se contaron los pedidos realizados en cada año.
#Recomendación: Analizar las causas de la baja cantidad de pedidos en ese año para identificar oportunidades de mejorar las ventas.
```

```python
#2 Cual es la linea de producto que mas se ha cancelado y cuantos fueron?
cancelados = (df["STATUS"] == "Cancelled") #filtro pra estado cancelado
dfCancelados = df.loc[cancelados] #loc para seleccionar las filas que tengan estado cancelado
productoCancelado = dfCancelados["PRODUCTLINE"].value_counts() #cuento los pedidos por categoria
print(productoCancelado.head(1))#head muesta la linea de producto mas cancelada y su cantidad
#Se encontró: La línea de producto que presenta con mayor cantidad de pedidos cancelados fue Ships con 18.
#Se comprobó: Se filtraron los pedidos con estado Cancelled y se contaron las apariciones de cada PRODUCTLINE.
#Recomendación: Analizar las causas de cancelación de esa línea para identificar posibles problemas y reducir las cancelaciones.
```

```python
#3 En que año se vendieron mas aviones y cuantos fueron?
aviones = (df["PRODUCTLINE"] == "Planes") # Creo el filtro para la línea de aviones
dfAviones = df.loc[aviones] # loc para seleccionar los pedidos de aviones
pedidosPorAnio = dfAviones["YEAR_ID"].value_counts() #Value_counts Cuenta los pedidos de cada año
print(pedidosPorAnio.head(1)) # head Muestro el año con más pedidos
# Se encontró: En el 2004 se vendieron mas aviones con 161.
# Se comprobó: Se filtraron los pedidos de la línea "Planes" y se contaron los pedidos de cada año.
# Recomendación: Comparar el año con mayor ventas con el actual para saber si han aumentado o disminuido las ventas de aviones
```

```python
#4: ¿En qué ciudades de Italia se realizaron ventas menores a $1,000?
italyMenosMil = (df["COUNTRY"] == "Italy") & (df["SALES"] < 1000) # Creo el filtro de pais y ventas menos a 1,000
dfItalyMenosMil = df.loc[italyMenosMil] # Selecciono los pedidos que cumplen el filtro
dfItalyMenosMilCiudades = dfItalyMenosMil[["CITY", "SALES"]] # Selecciono las columnas de ciudad y ventas
print(dfItalyMenosMilCiudades) # Muestro las ciudades y sus ventas menores a $1,000
#Se encontró: Las ciudades de Italia donde se realizaron ventas menores a $1,000.
#Se comprobó: Se filtraron los pedidos de Italia con SALES menores a $1,000 y se mostraron sus ciudades.
#Recomendación: Analizar las ventas de estas ciudades para identificar oportunidades de incrementar el valor de los pedidos.
```
