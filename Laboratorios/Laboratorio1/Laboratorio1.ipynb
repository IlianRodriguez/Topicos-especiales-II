```python
#1: Descargar Dataset
import pandas as pd #importo libreria panadas

url = "https://github.com/IlianRodriguez/Topicos-especiales-II/raw/refs/heads/main/Laboratorios/Laboratorio1/sales_data_sample.csv" #url del csv

df = pd.read_csv(url, encoding="cp1252") #leo el csv y lo guardo
```

```python
#2: Analisis exploratorio
print(df.shape) #muestra filas y columnas
print(df.columns) #muestra los nombres de las columnas
print(df.dtypes) #muestra el tipo de dato de cada columna
print(df.head()) #muestra las primeras 5 filas
print(df.tail()) #muestra las ultimas 5 filas
print(df.describe()) #muestra estadisticas básicas
print(df.info()) #muestra información general del DataFrame
print(df.isnull().sum()) #cuenta los valores nulos por columna
```

```python
#3: nuevo DataFrame con: ORDERNUMBER, PRODUCTLINE, COUNTRY, CUSTOMERNAME, QUANTITYORDERED, PRICEEACH, SALES, DEALSIZE.
columnas = ["ORDERNUMBER", "PRODUCTLINE", "COUNTRY", "CUSTOMERNAME", "QUANTITYORDERED", "PRICEEACH", "SALES", "DEALSIZE"] #Almaceno las columanas en una variable
df2 = df[columnas] #selecciono las columnas y las guardo en el nuevo dataset
print(df2.columns)
```
