## Nulos

```python

print("\n------------------ (Nulos) --------------------")
cantidadTotal = len(reservas)
cantidadNulos = reservas.isna().sum()
porcentajeNulos = (cantidadNulos / cantidadTotal * 100).round(4)

print(pd.DataFrame({
    "Total/Regsitros": cantidadTotal,
    "Total/Nulos": cantidadNulos,
    "(%)/Nulos": porcentajeNulos
}).loc[cantidadNulos > 0])
print("--------------------------------------------------")
```

Hallazgos  
* Se encontraron valores nulos en las columnas:
  1. children ----- (4,  0.0034%)
  2. country ----- (488, 0.4087%)
  3. agent   ------- (16340, 13.6862%)
  4. company ---- (112563, 94.3069%)  

Desisciones
  1. children: Se imputaron los valores nulos con la mediana (0), debido a que el sesgo de 4.11 supera el límite de 0.5 establecido.
  2. country: segun la regla del negocio, los hoteles pueden desconocer la nacionalidad de sus huespedes hasta el momento de su check-in, por lo su valor se reemplaza por "Desconocido"    
  3. agent: Segun la regla del negocio, los valores NULL significan que la reserva fue realizada sin un agente de viaje, por lo que su valor se reemplazara por "No aplicable".  
  4. company: Segun la regla del negocio, los valores NULL significan que la reserva fue realizada sin una agencia de viajes, por lo que su valor se reemplazara por "No aplicable".  


```python
nulosNoAplicable = ["agent", "company"]#Nulos no aplicables para agent y company
nulosDesconocido = ["country"]#Nulos desconocidos para country


for columna in nulosNoAplicable:
    reservas[columna] = reservas[columna].cat.add_categories(["No aplicable"])
    reservas[columna] = reservas[columna].fillna("No aplicable")

for columna in nulosDesconocido:
    reservas[columna] = reservas[columna].cat.add_categories(["Desconocido"])
    reservas[columna] = reservas[columna].fillna("Desconocido")

#Nulos de children según el sesgo
columna = "children"

media = reservas[columna].mean()
mediana = reservas[columna].median()
sesgo = reservas[columna].skew()

if abs(sesgo) < 0.5:
    valorImputacion = media
    criterio = "media"
else:
    valorImputacion = mediana
    criterio = "mediana"

reservas[columna] = reservas[columna].fillna(valorImputacion).astype("Int64")

print(f"{columna}: sesgo={sesgo:.2f} -> se imputó con {criterio} = {valorImputacion:.2f}")
```

```python
print("\n------------------ (Validacion de Nulos) --------------------")
cantidadNulos = reservas.isna().sum()
porcentajeNulos = (cantidadNulos / len(reservas) * 100).round(4)
if cantidadNulos.sum() > 0:
    print(pd.DataFrame({
        "Cantidad": cantidadNulos,
        "%": porcentajeNulos,
    }).loc[cantidadNulos > 0])
else:
    print("Mensaje: No hay valores nulos")
print("--------------------------------------------------")
```

# 6. media, mediana, desviación estándar, asimetría y coeficiente de variación.

---

```python
# Calcular el total de noches
reservas["total_noches"] = (
    reservas["stays_in_weekend_nights"]
    + reservas["stays_in_week_nights"]
)

# Seleccionar las variables
columnasNumericas = ["lead_time", "adr", "total_noches"]

for col in columnasNumericas:

    datos = reservas[col].dropna()

    mediaCol = datos.mean()
    medianaCol = datos.median()
    desviacion = datos.std()
    asimetria = datos.skew()
    coefVariacion = desviacion / mediaCol * 100 if mediaCol != 0 else float("nan")

    if abs(asimetria) < 0.5:
        criterio = "Media"
        valorRepresentativo = mediaCol
    else:
        criterio = "Mediana"
        valorRepresentativo = medianaCol

    print(f"\n---------- {col.upper()} ----------")
    print(f"Media: {mediaCol:.2f}")
    print(f"Mediana: {medianaCol:.2f}")
    print(f"Desviación estándar: {desviacion:.2f}")
    print(f"Asimetría: {asimetria:.2f}")
    print(f"Coeficiente de variación: {coefVariacion:.2f} %")
    print(f"Medida seleccionada: {criterio}")
    print(f"Valor representativo: {valorRepresentativo:.2f}")
```

Justificacion
  1. lead_time: la mediana de 69 días representa mejor el valor central, porque la asimetría positiva de 1.35 indica que existen reservas con anticipaciones mucho mayores que otras, elevando la media hasta 104.01 días.
  2. adr: la mediana de 94.58 representa mejor el valor central, porque la asimetría de 10.53 es muy elevada. Esto indica una distribución fuertemente sesgada hacia la derecha, posiblemente influida por tarifas excepcionalmente altas.
  3. total_noches: la mediana de 3 noches representa mejor el valor central, porque la asimetría de 3.31 indica que hay estancias largas que elevan la media hasta 3.43 noches.

Observacion final  
  1. En las tres variables se seleccionó la mediana porque presentan una asimetría positiva, lo que indica que existen valores elevados que alargan la distribución hacia la derecha. Estos valores extremos pueden aumentar la media y hacer que se aleje del valor central de la mayoría de los datos. Por esta razón, la mediana resulta más representativa, ya que es menos sensible a los valores extremos y refleja mejor el comportamiento central de las reservas.

# 9. Hipótesis

---



## 1. Planteamiento de la hipótesis

"Las reservaciones de los hoteles Resort Hotel y City Hotel que registran al menos un pedido especial presentan una tasa de cancelación menor que las reservaciones sin pedidos especiales, durante el año 2017."

---

## 2. Criterio de refutación

La hipótesis se refuta si las reservaciones de los hoteles Resort Hotel y City Hotel que registran al menos un pedido especial presentan una tasa de cancelación igual o superior a la de las reservaciones sin pedidos especiales, durante el año 2017.

---

## 3. Exploración de los datos de 2015 y 2016

```python
reservasExploracion = reservas[reservas["arrival_date_year"].isin([2015, 2016])].copy()# Filtrar los datos de 2015 y 2016

# Crear una columna para identificar los grupos
reservasExploracion["grupo_pedidos"] = "Sin pedidos especiales"
reservasExploracion.loc[reservasExploracion["total_of_special_requests"] >= 1,"grupo_pedidos"] = "Con pedidos especiales"

reservasExploracion["tasa_cancelacion"] = reservasExploracion["is_canceled"] * 100# Calcular la tasa de cancelación por reservación

pedidosEspecialesCancelacion = reservasExploracion.pivot_table(# Crear la tabla pivote
    index=["arrival_date_year", "grupo_pedidos"],
    values=["is_canceled", "tasa_cancelacion"],
    aggfunc={
        "is_canceled": ["count", "sum"],
        "tasa_cancelacion": "mean"
    }
)

# Cambiar los nombres de las columnas
pedidosEspecialesCancelacion.columns = ["total_cancelaciones","total_reservaciones","(%) tasa_cancelacion"]

display(pedidosEspecialesCancelacion.round(2))
```

Hallazgos  
Los resultados respaldan la hipótesis en ambos años: las reservaciones con al menos un pedido especial presentan una tasa de cancelación menor que las reservaciones sin pedidos especiales.
* Año 2015  
  - Con pedidos especiales: tasa de cancelación de 20.80 %.
  - Sin pedidos especiales: tasa de cancelación de 44.09 %.

Esto representa una diferencia de 23.33 puntos porcentuales. Por lo tanto, los resultados respaldan la hipótesis planteada, ya que las reservaciones con pedidos especiales presentaron una menor tasa de cancelación durante el año 2017.

```python
reservas2017 = reservas[reservas["arrival_date_year"] == 2017].copy()# Filtrar los datos de 2017
reservas2017 = reservas2017[["total_of_special_requests", "is_canceled"]].corr(method="spearman")# Calcular la correlación de Spearman
display(reservas2017.round(2))
```

## 5. Evaluación del efecto de los duplicados

```python
reservas_Con_Duplicados = reservas.copy()# Datos con duplicados
reservas_Sin_Duplicados = reservas.copy().drop_duplicates()# Datos sin duplicados exactos

# Crear los grupos con duplicados
reservas_Con_Duplicados["grupo_pedidos"] = "Sin pedidos especiales"
reservas_Con_Duplicados.loc[
    reservas_Con_Duplicados["total_of_special_requests"] >= 1,
    "grupo_pedidos"
] = "Con pedidos especiales"

# Crear los grupos sin duplicados
reservas_Sin_Duplicados["grupo_pedidos"] = "Sin pedidos especiales"
reservas_Sin_Duplicados.loc[
    reservas_Sin_Duplicados["total_of_special_requests"] >= 1,
    "grupo_pedidos"
] = "Con pedidos especiales"

# Tabla pivote con duplicados
resultado_Con_Duplicados = reservas_Con_Duplicados.pivot_table(
    index=["arrival_date_year", "grupo_pedidos"],
    values="is_canceled",
    aggfunc=["count", "sum", "mean"]
)

resultado_Con_Duplicados.columns = [
    "Total reservaciones",
    "Tasa de cancelación (%)",
    "Total cancelaciones"
]
resultado_Con_Duplicados["Tasa de cancelación (%)"] *= 100

print("---------------------------------------- (Evaluacion con Duplicados) ----------------------------------------")
display(resultado_Con_Duplicados.round(2))
print("-------------------------------------------------------------------------------------------------------------")

# Tabla pivote sin duplicados
resultado_Sin_Duplicados = reservas_Sin_Duplicados.pivot_table(
    index=["arrival_date_year", "grupo_pedidos"],
    values="is_canceled",
    aggfunc=["count", "sum", "mean"]
)

resultado_Sin_Duplicados.columns = [
    "Total reservaciones",
    "Tasa de cancelación (%)",
    "Total cancelaciones"
]
resultado_Sin_Duplicados["Tasa de cancelación (%)"] *= 100

print("\n---------------------------------------- (Evaluacion sin Duplicados) ----------------------------------------")
display(resultado_Sin_Duplicados.round(2))
print("-------------------------------------------------------------------------------------------------------------")
```

Hallazgos  
Tanto con duplicados como sin ellos, las reservaciones con pedidos especiales presentan tasas de cancelación menores que las reservaciones sin pedidos especiales en los años 2015, 2016 y 2017. Sin embargo, al eliminar los duplicados, las tasas de cancelación cambian, especialmente en el grupo sin pedidos especiales. Esto demuestra que los duplicados influyen en los resultados, aunque la tendencia general se mantiene.

```python
print("----------------------------------- (Evaluacion con Duplicados 2015-2016) ------------------------------------")
display(
    reservas_Con_Duplicados[
        reservas_Con_Duplicados["arrival_date_year"].isin([2015, 2016])
    ][["total_of_special_requests", "is_canceled"]]
    .corr(method="spearman").round(2)
)
print("----------------------------------- (Evaluacion con Duplicados 2017) ------------------------------------")
# Spearman con duplicados: 2017
print("Spearman con duplicados - Año 2017")
display(
    reservas_Con_Duplicados[
        reservas_Con_Duplicados["arrival_date_year"] == 2017
    ][["total_of_special_requests", "is_canceled"]]
    .corr(method="spearman").round(2)
)
print("-------------------------------------------------------------------------------------------------------------")

print("\n----------------------------------- (Evaluacion sin Duplicados 2015-2016) ------------------------------------")
display(
    reservas_Sin_Duplicados[
        reservas_Sin_Duplicados["arrival_date_year"].isin([2015, 2016])
    ][["total_of_special_requests", "is_canceled"]]
    .corr(method="spearman").round(2)
)
print("----------------------------------- (Evaluacion sin Duplicados 2017) ------------------------------------")

display(
    reservas_Sin_Duplicados[
        reservas_Sin_Duplicados["arrival_date_year"] == 2017
    ][["total_of_special_requests", "is_canceled"]]
    .corr(method="spearman").round(2)
)
print("-------------------------------------------------------------------------------------------------------------")
```

Hallazgos  
En todos los periodos se observa una relación negativa débil entre la cantidad de pedidos especiales y las cancelaciones. Al eliminar los duplicados, la correlación disminuye en magnitud, pasando de −0.28 a −0.13 en 2015-2016 y de −0.24 a −0.13 en 2017. Esto indica que los duplicados influyen en la intensidad de la relación observada, aunque la tendencia negativa se mantiene.
